# K-Means Clustering

1. **Algorithm** - Iterative centroid assignment and update
2. **Choosing K** - Elbow method and silhouette analysis
3. **Limitations** - When K-Means fails

**Dataset**: Customer segmentation with sklearn's make_blobs + Iris

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs, load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score, silhouette_samples

sns.set_theme(style="whitegrid")
```

```python
# Generate synthetic data
X, y_true = make_blobs(n_samples=500, centers=4, cluster_std=0.8, random_state=42)

# K-Means clustering
kmeans = KMeans(n_clusters=4, n_init=10, random_state=42)
labels = kmeans.fit_predict(X)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].scatter(X[:, 0], X[:, 1], c=y_true, cmap="viridis", s=20, alpha=0.6)
axes[0].set_title("True Labels")

axes[1].scatter(X[:, 0], X[:, 1], c=labels, cmap="viridis", s=20, alpha=0.6)
axes[1].scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1],
                c="red", marker="X", s=200, edgecolors="black", label="Centroids")
axes[1].set_title("K-Means Clusters")
axes[1].legend()

plt.tight_layout()
plt.show()
```

## Choosing K: Elbow Method + Silhouette Analysis

```python
K_range = range(2, 10)
inertias = []
silhouettes = []

for k in K_range:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    km.fit(X)
    inertias.append(km.inertia_)
    silhouettes.append(silhouette_score(X, km.labels_))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(K_range, inertias, "o-", color="teal")
axes[0].set_xlabel("K")
axes[0].set_ylabel("Inertia (within-cluster sum of squares)")
axes[0].set_title("Elbow Method")

axes[1].plot(K_range, silhouettes, "o-", color="coral")
axes[1].set_xlabel("K")
axes[1].set_ylabel("Silhouette Score")
axes[1].set_title("Silhouette Analysis")

plt.tight_layout()
plt.show()

print(f"Best K by silhouette: {K_range[np.argmax(silhouettes)]}")
```

## Real-World Example: Iris Clustering

```python
iris = load_iris()
X_iris = StandardScaler().fit_transform(iris.data)

km_iris = KMeans(n_clusters=3, n_init=10, random_state=42)
iris_labels = km_iris.fit_predict(X_iris)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].scatter(X_iris[:, 0], X_iris[:, 1], c=iris.target, cmap="viridis", s=30)
axes[0].set_title("True Species")
axes[0].set_xlabel(iris.feature_names[0])
axes[0].set_ylabel(iris.feature_names[1])

axes[1].scatter(X_iris[:, 0], X_iris[:, 1], c=iris_labels, cmap="viridis", s=30)
axes[1].set_title(f"K-Means (silhouette={silhouette_score(X_iris, iris_labels):.3f})")
axes[1].set_xlabel(iris.feature_names[0])
axes[1].set_ylabel(iris.feature_names[1])

plt.tight_layout()
plt.show()
```

## Key Takeaways

1. **K-Means assumes spherical, equal-size clusters** - fails on elongated or varying-density clusters
2. **Always scale features** before clustering
3. **Use elbow + silhouette together** to choose K - neither alone is conclusive
4. **Run multiple initializations** (n_init) to avoid bad local optima
5. **K-Means is fast** - O(nKd) per iteration, works well up to millions of points
